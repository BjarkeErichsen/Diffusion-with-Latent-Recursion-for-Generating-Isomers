# Molecular Conformer Generation, Evaluation, and Representation Probing
This notebook demonstrates loading a trained equivariant diffusion model from a snapshot, registering representation probes to capture internal activations over time, generating a new conformer in a single diffusion pass, evaluating its stability, and exporting the captured states as CSVs and visualization plots.

### Configuration Block
Customize the paths, validation molecule index, conditioning vector, and generation steps here.

```python
# ==============================================================================
# CONFIGURATION
# ==============================================================================
# Path to the snapshot directory (to load the corresponding model code)
SNAPSHOT_DIR = "../snapshots/GEOM_cfg_self_cond_s_v_edge_gated_rbf_run_20260608_024831"

# Path to the specific checkpoint weight file (.ckpt)
CHECKPOINT_PATH = "../snapshots/GEOM_cfg_self_cond_s_v_edge_gated_rbf_run_20260608_024831/output/geomdrugs/runs/2026-06-08_03-04-20/checkpoints/last.ckpt"

# Path to the validation dataset file
VAL_DATA_PATH = "../data/geom_drugs/preprocessed/val.pt"

# Index of the validation datapoint to use for chemical formula
VAL_DATA_INDEX = 0

# 3-dimensional conditioning vector (normalized eigenvalues)
# Set to None to use the validation molecule's actual eigenvalues_normalized
CONDITIONAL_VECTOR = None #[0.5, 0.5, 0]

# The property name of eigenvalues to use from target_data when CONDITIONAL_VECTOR is None
# Options: 'eigenvalues_normalized' or 'eigenvalues_normalized_no_h'
CONDITIONAL_EIGENVALUE_PROPERTY = "eigenvalues_normalized_no_h"

# Number of integration steps for diffusion sampling
N_INTEGRATION_STEPS = 250

# Classifier-free guidance scale
CFG_SCALE = 1.0

# Whether to visualize Hydrogen atoms and their bonds in the 3D plot
VISUALIZE_HYDROGENS = False

# Probes to capture during molecule generation
PROBES_TO_CAPTURE = [
    # Encoder sc-pre/post probes (including older before_sc naming)
    "parameterization.encoder.probes.pre_sc_s",
    "parameterization.encoder.probes.pre_sc_e",
    "parameterization.encoder.probes.post_sc_s",
    "parameterization.encoder.probes.post_sc_v",
    "parameterization.encoder.probes.post_sc_e",

    #position
    "parameterization.encoder.probes.prev_pos",
    "parameterization.encoder.probes.pos",
    
    # Encoder layer probes (s and v across all 8 layers)
    "parameterization.encoder.probes.layer_0_s",
    "parameterization.encoder.probes.layer_1_s",
    "parameterization.encoder.probes.layer_2_s",
    "parameterization.encoder.probes.layer_3_s",
    "parameterization.encoder.probes.layer_4_s",
    "parameterization.encoder.probes.layer_5_s",
    "parameterization.encoder.probes.layer_6_s",
    "parameterization.encoder.probes.layer_7_s",
    "parameterization.encoder.probes.layer_0_v",
    "parameterization.encoder.probes.layer_1_v",
    "parameterization.encoder.probes.layer_2_v",
    "parameterization.encoder.probes.layer_3_v",
    "parameterization.encoder.probes.layer_4_v",
    "parameterization.encoder.probes.layer_5_v",
    "parameterization.encoder.probes.layer_6_v",
    "parameterization.encoder.probes.layer_7_v",

    # Self-conditioning gate residual layer probes
    "parameterization.encoder.sc_layer.probes.sc_s_gate",
    "parameterization.encoder.sc_layer.probes.sc_edge_gate",
    "parameterization.encoder.sc_layer.probes.sc_v_gate"
]
# ==============================================================================
```

### 1. Imports and Environment Setup
Set the environment variables so that hydra configurations resolve properly, and insert the snapshot path into `sys.path` to load the code from the snapshot directory.

```python
import os
import sys
import numpy as np
import torch
import matplotlib.pyplot as plt
from torch_geometric.data import Batch
import hydra
from src_gmmm.utils import utils
from src_gmmm.lit.module import LitModule
from src_gmmm.data.dataset import Dataset
from src_gmmm.metrics.geom_drugs import GeomDrugsMetrics
from rdkit import Chem
from rdkit.Chem import rdDetermineBonds
from rdkit.Geometry import Point3D

# --- BULLETPROOF PATH RESOLUTION ---
# If we are running inside 'scripts/', move to the project root automatically
if os.path.basename(os.getcwd()) == 'scripts':
    os.chdir('..')

# Ensure the configured paths are absolute relative to the project root
project_root = os.getcwd()

if not SNAPSHOT_DIR.startswith('/'):
    SNAPSHOT_DIR = os.path.join(project_root, SNAPSHOT_DIR.replace('../', ''))
if not CHECKPOINT_PATH.startswith('/'):
    CHECKPOINT_PATH = os.path.join(project_root, CHECKPOINT_PATH.replace('../', ''))
if not VAL_DATA_PATH.startswith('/'):
    VAL_DATA_PATH = os.path.join(project_root, VAL_DATA_PATH.replace('../', ''))

os.environ['DATA_PATH'] = os.path.join(project_root, 'data')
os.environ['LOG_PATH'] = os.path.join(project_root, 'logs')

# --- FORCE LOADING FROM SNAPSHOT ---
snapshot_path = os.path.abspath(SNAPSHOT_DIR)
# Clean up any cached src_gmmm modules to force reload from the new snapshot path
for key in list(sys.modules.keys()):
    if key == 'src_gmmm' or key.startswith('src_gmmm.'):
        del sys.modules[key]

if snapshot_path in sys.path:
    sys.path.remove(snapshot_path)
sys.path.insert(0, snapshot_path)
```

### 2. Load Configuration and Target Molecule
We load the snapshot's job configuration and validation dataset, and retrieve the molecule at the chosen validation index to use as the formula target.

```python
import os
from pathlib import Path

# Ensure the checkpoint path is absolute and correctly resolved regardless of Jupyter CWD
if not os.path.exists(CHECKPOINT_PATH):
    if os.path.exists(os.path.join("..", CHECKPOINT_PATH)):
        CHECKPOINT_PATH = os.path.join("..", CHECKPOINT_PATH)
    elif "scripts/snapshots" in str(Path(CHECKPOINT_PATH).resolve()):
        # Handle the case where CWD is scripts/ and CHECKPOINT_PATH is snapshots/...
        CHECKPOINT_PATH = CHECKPOINT_PATH.replace("scripts/", "")
        if not os.path.exists(CHECKPOINT_PATH):
            CHECKPOINT_PATH = os.path.join("..", CHECKPOINT_PATH)

CHECKPOINT_PATH = os.path.abspath(CHECKPOINT_PATH)
if not os.path.exists(CHECKPOINT_PATH):
    raise FileNotFoundError(f"Checkpoint path does not exist: {CHECKPOINT_PATH}")

print(f"Loading job configuration from: {CHECKPOINT_PATH}")
job_dir, cfg = utils.load_cfg(checkpoint=CHECKPOINT_PATH)

# Overwrite hydra runtime paths to prevent resolution failure outside of training
if "output_dir" in cfg:
    cfg.output_dir = job_dir
if "work_dir" in cfg:
    cfg.work_dir = job_dir

print("Job configuration loaded successfully.")

MODEL_HAS_HYDROGENS = not getattr(cfg, "remove_h", False)
print(f"Model models hydrogen: {MODEL_HAS_HYDROGENS}")
```

### 3. Load Trained Model
Instantiate the equivariant diffusion model and load the checkpoint weights.

```python
# Instantiate Lightning Module and load weights
print("Instantiating LitModule...")
device = torch.device("cuda") if torch.cuda.is_available() else torch.device("cpu")
print(f"Using device: {device}")

lit_module = hydra.utils.instantiate(cfg.lit_module)
print(f"Loading checkpoint weights from: {CHECKPOINT_PATH}")
ckpt = torch.load(CHECKPOINT_PATH, map_location=device)
lit_module.load_state_dict(ckpt["state_dict"], strict=False)
lit_module.to(device)
lit_module.eval()

# Override integration steps and CFG scale
lit_module.hparams.n_integration_steps = N_INTEGRATION_STEPS
lit_module.hparams.cfg_scale = CFG_SCALE
```

### 4. Setup Probes & Run Generation (Single Pass)
We register hooks on the active model to capture internal representations over all generation timesteps, and then execute the diffusion generation process exactly once.

```python
# 1. Identify active model and register hooks
model_to_probe = lit_module.get_model(ema=True)
model_modules = dict(model_to_probe.named_modules())

# We will collect activations over time
captured_temporal_tensors = {name: [] for name in PROBES_TO_CAPTURE}

def make_hook(probe_name):
    def hook(module, input, output):
        captured_temporal_tensors[probe_name].append(output.detach().cpu())
    return hook

hooks = []
for name in PROBES_TO_CAPTURE:
    if name in model_modules:
        hook_fn = make_hook(name)
        hooks.append(model_modules[name].register_forward_hook(hook_fn))
        print(f"Registered hook for: {name}")
    else:
        print(f"WARNING: Probe {name} not found in model.")

# 2. Load the validation dataset and select target
print(f"\nLoading validation dataset from: {VAL_DATA_PATH}")
val_dataset = torch.load(VAL_DATA_PATH)
target_data = val_dataset[VAL_DATA_INDEX]
print(f"Target molecule (index {VAL_DATA_INDEX}): {target_data.smiles}")

if "transform" in cfg.datamodule and cfg.datamodule.transform is not None:
    transform = hydra.utils.instantiate(cfg.datamodule.transform)
    target_data = transform(target_data)

batch = Batch.from_data_list([target_data])

if CONDITIONAL_VECTOR is None:
    # Dynamically select the eigenvalue property based on whether the model models hydrogen
    prop_name = "eigenvalues_normalized" if MODEL_HAS_HYDROGENS else "eigenvalues_normalized_no_h"
    # Support user's alternative naming 'eigenvalues_no_h_normalized' or 'eigenvalues_normalized_no_h'
    if not hasattr(target_data, prop_name):
        if prop_name == "eigenvalues_no_h_normalized" and hasattr(target_data, "eigenvalues_normalized_no_h"):
            prop_name = "eigenvalues_normalized_no_h"
        elif prop_name == "eigenvalues_normalized_no_h" and hasattr(target_data, "eigenvalues_no_h_normalized"):
            prop_name = "eigenvalues_no_h_normalized"
    
    if hasattr(target_data, prop_name):
        c_vector = getattr(target_data, prop_name).clone()
        print(f"Conditioning vector: using target's {prop_name} -> {c_vector.tolist()}")
    else:
        c_vector = target_data.eigenvalues_normalized.clone()
        print(f"Conditioning vector: using target's eigenvalues_normalized (fallback) -> {c_vector.tolist()}")
else:
    c_vector = torch.tensor(CONDITIONAL_VECTOR, dtype=torch.float32)
    print(f"Conditioning vector: using provided vector -> {c_vector.tolist()}")
    
cfg_property_name = getattr(cfg, "cfg_property", "eigenvalues_normalized")
setattr(batch, cfg_property_name, c_vector.view(1, -1).to(device))

# 3. Generate conformer (runs model forward pass internally)
print("\nGenerating conformer coordinates (single pass)...")
with torch.no_grad():
    atoms_list = lit_module.sample(batch.to(device), ema=True)
generated_atoms = atoms_list[0]
print("Generation complete!")

# 4. Remove hooks immediately
for hook in hooks:
    hook.remove()
print("Hooks removed successfully.")

# 5. Extract conditional states (handles CFG double forward calls)
final_temporal_states = {}
for name, tensors in captured_temporal_tensors.items():
    if len(tensors) == 0:
        continue
    if lit_module.model.cfg:
        # CFG runs unconditional first (even), conditional second (odd)
        cond_tensors = [tensors[idx] for idx in range(1, len(tensors), 2)]
        final_temporal_states[name] = cond_tensors
    else:
        final_temporal_states[name] = tensors
```

### 5. Evaluate Stability Locally
Compute stability statistics using local `GeomDrugsMetrics` (RDKit valence table and sanitization checks).

```python
print("\nEvaluating stability of the generated conformer...")
metrics = GeomDrugsMetrics(
    atom_types_str=cfg.atom_types_str,
    max_num_atoms=cfg.max_num_atoms,
    json_path=cfg.infos_path,
    remove_h=cfg.remove_h
)
metrics.update(generated_atoms)
summary = metrics.summarize()

print("\n" + "="*50)
print("STABILITY METRICS SUMMARY")
print(f"  Valid RDKit structure determined: {metrics.valid[0] == 1}")
print(f"  Single connected component:      {metrics.valid_connected[0] == 1}")
print(f"  Atom stable ratio:               {summary.get('atom_stable', 0.0):.4f}")
print(f"  Molecule stable:                 {summary.get('molecule_stable', 0.0) == 1.0}")
print(f"  Valency table molecule stable:   {summary.get('valency_table_molecule_stable', 0.0) == 1.0}")
print(f"  Generated SMILES:                {metrics.smiles[0]}")
print("="*50 + "\n")
```

### 6. Crisp 3D Conformer Visualization
Visualize the generated conformer in a high-quality, crisp 3D plot. The plotting code maps atomic elements to their specific standard colors, labels each atom with its chemical symbol, draws RDKit-determined bonds in solid black lines, and conditionally draws bonds with hydrogen atoms if enabled.

```python
import numpy as np
import plotly.graph_objects as go
from rdkit import Chem
from rdkit.Chem import rdDetermineBonds
from rdkit.Geometry import Point3D

# Assuming generated_atoms.symbols, generated_atoms.positions, 
# and VISUALIZE_HYDROGENS are already defined in your environment
symbols = generated_atoms.symbols
positions = generated_atoms.positions

def plot_molecule_conformer(mol, visualize_hydrogens=True, title="Generated 3D Conformer"):
    symbols = [atom.GetSymbol() for atom in mol.GetAtoms()]
    positions = mol.GetConformer().GetPositions()
    
    color_map = {
        'H': 'white', 'C': 'gray', 'N': 'blue', 'O': 'red', 'F': 'green', 
        'P': 'orange', 'S': 'yellow', 'Cl': 'darkgreen', 'Br': 'brown', 'I': 'purple'
    }
    size_map = {
        'H': 8, 'C': 12, 'N': 12, 'O': 12, 'F': 12, 
        'P': 14, 'S': 14, 'Cl': 14, 'Br': 15, 'I': 16
    }

    # 1. Prepare Atom Data
    indices_to_plot = [i for i, sym in enumerate(symbols) if sym != 'H' or visualize_hydrogens]

    atom_x = [positions[i, 0] for i in indices_to_plot]
    atom_y = [positions[i, 1] for i in indices_to_plot]
    atom_z = [positions[i, 2] for i in indices_to_plot]
    atom_symbols = [symbols[i] for i in indices_to_plot]

    atom_colors = [color_map.get(sym, 'magenta') for sym in atom_symbols]
    atom_sizes = [size_map.get(sym, 10) for sym in atom_symbols]
    text_colors = ["black" if sym == 'H' else "white" for sym in atom_symbols]

    # 2. Prepare Bond & Distance Label Data
    heavy_bonds_x, heavy_bonds_y, heavy_bonds_z = [], [], []
    h_bonds_x, h_bonds_y, h_bonds_z = [], [], []

    bond_mid_x, bond_mid_y, bond_mid_z = [], [], []
    bond_labels = []

    for bond in mol.GetBonds():
        idx1 = bond.GetBeginAtomIdx()
        idx2 = bond.GetEndAtomIdx()
        sym1 = symbols[idx1]
        sym2 = symbols[idx2]
        
        is_h_bond = (sym1 == 'H' or sym2 == 'H')
        if is_h_bond and not visualize_hydrogens:
            continue
            
        p1 = positions[idx1]
        p2 = positions[idx2]
        
        # Calculate distance and midpoint for the label
        dist = np.linalg.norm(p1 - p2)
        mid = (p1 + p2) / 2.0
        
        bond_mid_x.append(mid[0])
        bond_mid_y.append(mid[1])
        bond_mid_z.append(mid[2])
        bond_labels.append(f"{dist:.2f} Å")
        
        if is_h_bond:
            h_bonds_x.extend([p1[0], p2[0], None])
            h_bonds_y.extend([p1[1], p2[1], None])
            h_bonds_z.extend([p1[2], p2[2], None])
        else:
            heavy_bonds_x.extend([p1[0], p2[0], None])
            heavy_bonds_y.extend([p1[1], p2[1], None])
            heavy_bonds_z.extend([p1[2], p2[2], None])

    # 3. Create Traces
    traces = []

    if heavy_bonds_x:
        traces.append(go.Scatter3d(
            x=heavy_bonds_x, y=heavy_bonds_y, z=heavy_bonds_z,
            mode='lines',
            line=dict(color='black', width=4),
            hoverinfo='none',
            name='Bonds'
        ))

    if h_bonds_x:
        traces.append(go.Scatter3d(
            x=h_bonds_x, y=h_bonds_y, z=h_bonds_z,
            mode='lines',
            line=dict(color='gray', width=2),
            hoverinfo='none',
            name='H-Bonds'
        ))

    # Trace for the distance labels
    if bond_labels:
        traces.append(go.Scatter3d(
            x=bond_mid_x, y=bond_mid_y, z=bond_mid_z,
            mode='text',
            text=bond_labels,
            textposition='middle center',
            textfont=dict(color='cyan', size=10, family="Arial"),
            hoverinfo='none',
            name='Distances'
        ))

    traces.append(go.Scatter3d(
        x=atom_x, y=atom_y, z=atom_z,
        mode='markers+text',
        marker=dict(
            size=atom_sizes,
            color=atom_colors,
            line=dict(color='black', width=2),
            opacity=1.0
        ),
        text=atom_symbols,
        textposition="middle center",
        textfont=dict(color=text_colors, size=11, family="Arial Black"),
        hoverinfo='text', 
        name='Atoms'
    ))

    # 4. Build and Show Figure
    fig = go.Figure(data=traces)

    fig.update_layout(
        title=title,
        scene=dict(
            xaxis_title="X (Å)",
            yaxis_title="Y (Å)",
            zaxis_title="Z (Å)",
            xaxis=dict(showgrid=True, zeroline=False, showbackground=False),
            yaxis=dict(showgrid=True, zeroline=False, showbackground=False),
            zaxis=dict(showgrid=True, zeroline=False, showbackground=False)
        ),
        showlegend=False,
        margin=dict(l=0, r=0, b=0, t=40)
    )

    fig.show()

# Reconstruct bonds using RDKit for the generated atoms
mol = Chem.RWMol()
for symbol in symbols:
    mol.AddAtom(Chem.Atom(symbol))
conf = Chem.Conformer(len(generated_atoms))
for idx, pos in enumerate(positions):
    conf.SetAtomPosition(idx, Point3D(float(pos[0]), float(pos[1]), float(pos[2])))
mol.AddConformer(conf)

try:
    rdDetermineBonds.DetermineBonds(mol)
except Exception as e:
    print(f"WARNING: Bond determination failed for plotting: {e}")

# Check if model has hydrogens (is trained with hydrogens)
if MODEL_HAS_HYDROGENS:
    # If the model has hydrogens, just plot the conformer.
    # We respect the VISUALIZE_HYDROGENS flag from the configuration.
    plot_molecule_conformer(
        mol, 
        visualize_hydrogens=VISUALIZE_HYDROGENS, 
        title=f"Generated Conformer (Model has hydrogens, VISUALIZE_HYDROGENS={VISUALIZE_HYDROGENS})"
    )
else:
    # If the model does not have hydrogens, we show two plots:
    # 1. Without hydrogens added
    print("Model was trained without hydrogens. Visualizing generated molecule without hydrogens:")
    plot_molecule_conformer(
        mol, 
        visualize_hydrogens=False, 
        title="Generated Conformer (Without Hydrogens)"
    )
    
    # 2. With hydrogens added via RDKit
    print("Visualizing generated molecule with RDKit-added hydrogens:")
    try:
        mol_with_hs = Chem.AddHs(mol, addCoords=True)
        plot_molecule_conformer(
            mol_with_hs, 
            visualize_hydrogens=True, 
            title="Generated Conformer (With RDKit-Added Hydrogens)"
        )
    except Exception as e:
        print(f"WARNING: Failed to add hydrogens for visualization: {e}")
```

### 7. Final-Step Representation Probes
We visualize the final timestep's activations for each tracked probe. Vectors (`v`) are reduced to their L2 norm across the spatial dimension, while scalars (`s`/`e`) show absolute values. No PCA is plotted.

```python
import seaborn as sns

def plot_final_activations(states_dict):
    num_plots = len(states_dict)
    if num_plots == 0:
        print("No activations captured.")
        return
        
    fig, axes = plt.subplots(num_plots, 2, figsize=(16, 5 * num_plots))
    if num_plots == 1:
        axes = [axes]
        
    for ax_row, (name, tensors_list) in zip(axes, states_dict.items()):
        ax_heat, ax_dist = ax_row
        
        # Take the final timestep representation (last element)
        tensor = tensors_list[-1]
        
        # Check if it's a vector feature (v)
        is_vector = (tensor.dim() == 3 and tensor.shape[1] == 3) or (name.endswith('_v'))
        if is_vector:
            # L2 norm across spatial dimensions (dim 1)
            tensor_plot = torch.norm(tensor, dim=1).numpy()
            title_suffix = "(Vector L2 Norm)"
        else:
            # Absolute values of the scalar features
            tensor_plot = torch.abs(tensor).numpy()
            title_suffix = "(Scalar Absolute Value)"
            
        # Squeeze and reshape to 2D
        if tensor_plot.ndim > 2:
            tensor_plot = tensor_plot.squeeze()
        if tensor_plot.ndim > 2:
            # Flatten dimensions after the first to ensure 2D
            tensor_plot = tensor_plot.reshape(tensor_plot.shape[0], -1)
        if tensor_plot.ndim == 1:
            tensor_plot = tensor_plot.reshape(-1, 1)
        elif tensor_plot.ndim == 0:
            tensor_plot = tensor_plot.reshape(1, 1)
            
        # Heatmap
        sns.heatmap(tensor_plot, cmap="viridis", cbar=True, ax=ax_heat)
        ax_heat.set_title(f"{name} Final Heatmap {title_suffix}")
        ax_heat.set_xlabel("Feature Dimension")
        ax_heat.set_ylabel("Node/Edge Index")
        
        # Value Distribution (Histogram)
        ax_dist.hist(tensor_plot.flatten(), bins=50, color="steelblue", edgecolor="black", alpha=0.7)
        ax_dist.set_title(f"{name} Value Distribution")
        ax_dist.set_xlabel("Absolute Value" if not is_vector else "L2 Norm Value")
        ax_dist.set_ylabel("Frequency")
        ax_dist.grid(True, alpha=0.3)
            
    plt.tight_layout(pad=3.0, h_pad=3.0)
    plt.show()

plot_final_final = {k: v for k, v in final_temporal_states.items()}
plot_final_activations(plot_final_final)
```

### 8. Temporal Probe Analysis & Export
We export the full temporal profiles to CSV files inside the `scripts/tmp` directory (which can be easily loaded into pandas) and export sequential heatmaps of the TARGET_PROBE to `scripts/tmp/plots_delete_later`.

```python
import pandas as pd
import plotly.graph_objects as go

# Define export paths (under scripts/tmp)
tmp_dir = "scripts/tmp"
plots_dir = os.path.join(tmp_dir, "plots_delete_later")
os.makedirs(tmp_dir, exist_ok=True)
os.makedirs(plots_dir, exist_ok=True)

PLOT_INTERVAL = 10
TARGET_PROBE = "parameterization.encoder.probes.post_sc_v"

"""
    # Encoder sc-pre/post probes (including older before_sc naming)
    "parameterization.encoder.probes.pre_sc_s",
    "parameterization.encoder.probes.pre_sc_e",
    "parameterization.encoder.probes.post_sc_s",
    "parameterization.encoder.probes.post_sc_v",
    "parameterization.encoder.probes.post_sc_e",

    "parameterization.encoder.probes.prev_pos",
    "parameterization.encoder.probes.pos",

"""

# 1. Export temporal representations to CSV (Absolute value / L2 Norm)
for name, step_tensors in final_temporal_states.items():
    rows = []
    for step_idx, tensor in enumerate(step_tensors):
        if step_idx % PLOT_INTERVAL != 0:
            continue
            
        is_vector = (tensor.dim() == 3 and tensor.shape[1] == 3) or (name.endswith('_v'))
        if is_vector:
            # L2 norm across spatial dimensions
            tensor_np = torch.norm(tensor, dim=1).numpy()
        else:
            # Absolute values
            tensor_np = torch.abs(tensor).numpy()
            
        # Squeeze and reshape to 2D
        if tensor_np.ndim > 2:
            tensor_np = tensor_np.squeeze()
        if tensor_np.ndim > 2:
            tensor_np = tensor_np.reshape(tensor_np.shape[0], -1)
        if tensor_np.ndim == 1:
            tensor_np = tensor_np.reshape(-1, 1)
        elif tensor_np.ndim == 0:
            tensor_np = tensor_np.reshape(1, 1)
            
        num_items, num_features = tensor_np.shape
        for item_idx in range(num_items):
            row = {
                "step": step_idx,
                "index": item_idx,
            }
            # Feature columns: f0, f1, f2, ...
            for f_idx in range(num_features):
                row[f"f{f_idx}"] = tensor_np[item_idx, f_idx]
            rows.append(row)
            
    # Save CSV
    df = pd.DataFrame(rows)
    clean_name = name.replace(".", "_")
    csv_path = os.path.join(tmp_dir, f"probe_{clean_name}.csv")
    df.to_csv(csv_path, index=False)
    print(f"Exported {name} to {csv_path} (Shape: {tensor_np.shape}, Rows: {len(df)})")

# 2. Export Temporal Heatmap Images for target probe
print(f"\nSaving temporal heatmap images for {TARGET_PROBE} to {plots_dir}...")
if TARGET_PROBE in final_temporal_states:
    step_tensors = final_temporal_states[TARGET_PROBE]
    for step_idx, tensor in enumerate(step_tensors):
        if step_idx % PLOT_INTERVAL != 0:
            continue
            
        is_vector = (tensor.dim() == 3 and tensor.shape[1] == 3) or (TARGET_PROBE.endswith('_v'))
        if is_vector:
            tensor_plot = torch.norm(tensor, dim=1).numpy()
            z_label = "L2 Norm"
        else:
            tensor_plot = torch.abs(tensor).numpy()
            z_label = "Absolute Value"
            
        # Squeeze and reshape to 2D
        if tensor_plot.ndim > 2:
            tensor_plot = tensor_plot.squeeze()
        if tensor_plot.ndim > 2:
            tensor_plot = tensor_plot.reshape(tensor_plot.shape[0], -1)
        if tensor_plot.ndim == 1:
            tensor_plot = tensor_plot.reshape(-1, 1)
        elif tensor_plot.ndim == 0:
            tensor_plot = tensor_plot.reshape(1, 1)
            
        fig = go.Figure(data=go.Heatmap(
            z=tensor_plot,
            colorscale="Viridis",
            colorbar=dict(title=z_label)
        ))
        fig.update_layout(
            title=f"<b>Probe:</b> {TARGET_PROBE}<br><b>Diffusion Step:</b> {step_idx}",
            xaxis_title="Feature Dimension",
            yaxis_title="Atom/Edge Index",
            width=800, height=600,
            template="plotly_white"
        )
        
        file_path = os.path.join(plots_dir, f"{TARGET_PROBE.replace('.', '_')}_step_{step_idx:04d}.png")
        fig.write_image(file_path)
    print("All temporal heatmap images successfully saved!")
else:
    print(f"WARNING: Target probe {TARGET_PROBE} not found in captured states.")
```
